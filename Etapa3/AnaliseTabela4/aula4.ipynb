```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path = r"H:\PYTHON\Etapa3\AnaliseTabela4\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"
```

```python
dados_brutos = pd.read_excel(path, header=2)
```

```python
print(dados_brutos)
```

```python
dados_limpos = dados_brutos.dropna(subset=["UF"])
```

```python
col_2010 = "População 2010 (Alterações de Limites até 2022)1"
col_2022 = "População Censo 2022"

dados_limpos[col_2010] = pd.to_numeric(dados_limpos[col_2010], errors='coerce')
dados_limpos[col_2022] = pd.to_numeric(dados_limpos[col_2022], errors='coerce')
```

```python
variavel_a_criar = (
    dados_limpos.groupby("UF")[[col_2010, col_2022]]
    .sum()
    .reset_index()
)

variavel_a_criar["Crescimento_Absoluto"] = variavel_a_criar[col_2022] - variavel_a_criar[col_2010]
tabela_estado_ordenada = variavel_a_criar.sort_values(by="Crescimento_Absoluto", ascending=False)

tabela_estado_ordenada.to_csv("populacao_estado_crescimento.csv", index=False)
```

```python
dados_limpos["Crescimento_Absoluto"] = dados_limpos[col_2022] - dados_limpos[col_2010]
tabela_municipio_ordenada = dados_limpos.sort_values(by="Crescimento_Absoluto", ascending=False)
```

```python
colunas_finais = ["UF", "NOME DO MUNICÍPIO", col_2010, col_2022, "Crescimento_Absoluto"]
tabela_municipio_final = tabela_municipio_ordenada[colunas_finais]
```

```python
tabela_municipio_final.to_csv("populacao_municipio_crescimento.csv", index=False)

print("Exercício concluído! Os arquivos CSV foram salvos na pasta do projeto.")
```
